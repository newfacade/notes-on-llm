# Concise Reasoning via Reinforcement Learning

```{note}
In this work, we revisit the
core principles of reinforcement learning (RL) and, through mathematical analysis,
demonstrate that the tendency to generate lengthy responses arises inherently from
RL-based optimization during training.<br>
We show that introducing a secondary phase of RL training,
using a very small set of problems, can significantly reduce chains of thought
while maintaining or even enhancing accuracy.
```

## Response Length vs. Accuracy

```{tip}
In both reasoning and non-reasoning models, with
or without RL training, <span style="color: red">brevity and accuracy are strongly correlated</span>. A related explanation for why long responses may reduce accuracy is the notion of deadends. In practice, deadends still occur in LLMs despite strong sampling
techniques, often appearing as endless repetition.
```

We propose a novel approach: enforcing conciseness through a subsequent phase of RL training
with a dataset of occasionally solvable problems.

1. Normal RL training.
2. Training continues on problems with non-zero $p_a$ (occasionally solvable).
This phase enforces conciseness while preserving or even enhancing accuracy.
