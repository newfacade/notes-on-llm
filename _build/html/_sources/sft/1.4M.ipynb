# 1.4 Million Open-Source Distilled Reasoning Dataset

```{note}
The AM-DeepSeek-R1-Distilled is a large-scale dataset with thinking traces for general reasoning tasks. These problems are collected from a multitude of open-source datasets, all responses within the dataset are distilled from reasoning models
(predominantly DeepSeek-R1) and have undergone <span style="color: red">rigorous verification procedures.</span>
```

```{figure} ../images/1.4m-1.png
```

## Raw Data

### Data Sources

We divided the data selection into four major categories: math, code, scienceQA, and general chat. For Math, Code, and ScienceQA, we
prioritized to select datasets with reference answers or test cases, such as `OpenCoder`, `PRIME`, `CodeIO`. Additionally, we also selected some datasets with reasoning chains generated by
DeepSeek-R1 from the open-source community, such as `Openthoughts`, `KodCode`.

### Difficulty

We used a large language model to
score the difficulty of all instructions, subsequently screening the data and downsampling easy and medium
difficulty examples.

### Deduplication

We calculated the embedding for each
data entry and computed text similarity based on their embeddings to obtain the semantic similarity of
different data entries. For data with high semantic similarity, we designed some priority strategies and
ultimately retained only one representative entry.

## Distilled Data

We obtained responses to prompts via two ways: filtering existing responses and creating new responses. For
prompts with existing responses, we retained the original response if it can pass reference-answer or test-case
verification. For data without reasoning chains, we generated new responses using DeepSeek-R1.

### Ground Truth Verification

For code-related problems with test cases, we
verified responses within a sandbox environment.

### Reward

We used two methods, Decision-Tree-Reward-Llama-3.1-8B as reward model and Qwen2.5-
7B-Instruct for large language model scoring.
