# Parallel

```{tip}
在分布式训练大模型时，fsdp 和 megatron 是两种常见的模型并行与分布式训练策略.

1. FSDP（Fully Sharded Data Parallel）
    * FSDP 是 PyTorch 原生支持的分布式训练策略，属于 “完全分片数据并行”。它将模型的参数、梯度和优化器状态（Optimizer States）全部分片到不同的 GPU 上，每个 GPU 仅保留部分模型数据，同时结合数据并行（Data Parallel）的思想进行训练。
    * 优势：
        * 内存效率高：适合超大规模模型（模型大小超过单卡内存）。
        * 兼容性好：原生支持 PyTorch 模型，无需大幅修改模型结构。
        * 灵活性强：可与模型并行（Model Parallel）结合使用，支持混合策略。
2. Megatron（基于 Megatron-LM 的策略）
    * Megatron 是 NVIDIA 提出的大模型训练框架（Megatron-LM）中的分布式策略，主要依赖张量并行（Tensor Parallelism）和流水线并行（Pipeline Parallelism）：
        * 张量并行：将模型的单个层（如注意力层、线性层）的权重张量拆分到多个 GPU 上，每个 GPU 计算层的部分结果，再通过通信聚合，适合加速单层计算。
        * 流水线并行：将模型的不同层分配到不同的 GPU 上，形成流水线式计算，每层处理一部分数据，提高计算效率。
    * 优势：
        * 计算效率高：针对大模型层内并行优化，适合超大规模模型的高效训练。
        * 支持极致扩展：可通过组合张量并行、流水线并行和数据并行，支持上万张 GPU 协同训练。
```
