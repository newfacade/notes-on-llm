# SFT Data

## Query

1. Human written (InstructGPT)
2. User query (InstructGPT)
3. Synthetic data from
   * Code-related Q&A websites (Qwen2.5, Seed-Coder)
   * Github (Qwen2.5, Qwen2.5-Coder, Seed-Coder)
   * Open-source instruction dataset, e.g McEval-Instruct(Qwen2.5-Coder)
   * Code snippets from various source
   * programming language translation
   * backtranslation
4. Competitive programming contests (Seed-1.5-Thinking)
5. Opensource dataset

## Filtering

1. Topic classification
    * Fine-tune model into a topic classifier (LLama3)
2. Quality scoring
    * Model based (LLama3)
    * RM score (LLama3)
4. Difficulty scoring
    * Instag (LLama3, Seed-Coder)
    * Model-based scoring (LLama3)
