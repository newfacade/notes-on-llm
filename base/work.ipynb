## Arena/日志评估集

方法：建设数据飞轮

* 评估：
    * 评估集 come from 日志数据（not in any other use），based on 数据画像
    * 人工评估（版本更新等）
    * 自动评估（模型快速调优，数据筛选）
    * `自动评估 match 人工评估`，prompt + LLM 选型
* query
    * 缺哪补哪
    * 合成数据（前端 magicoder）
    * 自动评估
    * IFD（Instruction-Following Difficulty）
    * Diversity based on Sentence Embedding
    * Difficulty 改数据分布
* response 刷新
    * 自动评估
    * 过滤：Tree-sitter 等
    * Rejection Sampling based 上一版最终模型
        * `通过 bleu score，发现 RFT 相比 SFT 更好学` (accuracy & bleu score)
    * Or better 竞品模型
    * Pre-SFT
        * 第一轮普通 SFT
        * 通过 bleu 值发现哪些 query-response 不好学，不好学的交给 RFT
        * Due to differences between the pre-training datasets used for these LLMs and our own data, the distillation process can introduce significant alignment "taxes."
        * To mitigate these issues, we often opt for RFT over distillation. Although RFT helps in reducing alignment taxes, it requires sampling multiple candidate responses from an SFT model, which is time-consuming and labor-intensive.
        * Pre-SFT 结合了两者的优点
