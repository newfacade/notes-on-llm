### 描述
合数是指自然数中除了能被1和本身整除外，还能被其他数（0除外）整除的数。    牛牛最近在研究“k合因子数”，所谓“k合数”是指一个数的所有因子中，是合数的因子共有k个。    例如20的因子有1,2,4,5,10,20，其中4,10,20为合数，它有3个合数因子，就称20是一个 “3合因子数”    牛牛想要知道1~n中给定k的情况下k合因子数的数目。
### 输入描述：
第一行输入两个数字n,m$ (1 \leq n,m \leq 10^5)$表示范围以及查询“k”的数目接下来m行，每行一个正整数k$ (1 \leq k \leq n)$查询k合因子数的数目。
### 输出描述：
一行一个数字，表示k合因子数的数目
### 示例1
输入：
10 5
1
2
3
4
5
输出：
4
1
0
0
0
说明：1~10的范围内1合因子数有：4,6,9,10，共4个2合因子数有：8，共1一个
